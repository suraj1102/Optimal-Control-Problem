```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import numpy as np
plt.style.use("tableau-colorblind10")
```

```python
OUTPUT_DIR = "outputs"
```

```python
df = pd.read_csv(f"{OUTPUT_DIR}/xtfc.csv")
```

```python
df.head()
```

```python
Q = np.diag([2, 0.1])
R = np.diag([0.5])
```

```python
import numpy as np

def compute_cost(row):
    x = np.array([row['state1'], row['state2']])
    u = np.array([row['u']])
    cost = x @ Q @ x + u @ R @ u
    return -cost

df['rewards'] = df.apply(compute_cost, axis=1)
df.head()
```

```python
# Rename columns to theta, thetadot, control, reward
df = df.rename(columns={
    'state1': 'theta',
    'state2': 'thetadot',
    'u': 'control',
    'rewards': 'reward'
})
df.head()
```

```python
# Save the DataFrame to CSV
df.to_csv(f"{OUTPUT_DIR}/xtfc_rollout.csv", index=False)
```
